  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432037_NiNyomanSutriani
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432037_NiNyomanSutriani
# 1. Muat dataset dengan upload terlebih dahulu
df = pd.read_csv('dataset_mbg.csv')
print('Jumlah data:', df.shape)
print(df['sentimen'].value_counts())

# 2. Pra-pemrosesan teks: case folding, pembersihan, tokenisasi
def clean_text(text):
    text = str(text).lower()                                   # case folding
    text = text.replace('&amp;', ' dan ').replace('&gt;', ' ').replace('&lt;', ' ')
    text = re.sub(r'https?://\S+', ' ', text)                  # hapus URL
    text = re.sub(r'@\w+', ' ', text)                          # hapus mention
    text = re.sub(r'[^a-z\s]', ' ', text)                      # hapus angka, emoji, tanda baca, simbol #
    return re.sub(r'\s+', ' ', text).strip()

df['teks_bersih'] = df['teks'].apply(clean_text)
df['tokens'] = df['teks_bersih'].str.split()                   # tokenisasi
df = df[df['tokens'].str.len() > 0].reset_index(drop=True)

# 3. Split data latih & uji (80:20, stratified agar proporsi kelas terjaga)
train_df, test_df = train_test_split(df, test_size=0.2, stratify=df['label'], random_state=SEED)

# 4. Vocabulary mapping (hanya dari data latih agar tidak ada kebocoran data)
MIN_FREQ = 2
counter = Counter(tok for toks in train_df['tokens'] for tok in toks)
vocab = {'<PAD>': 0, '<UNK>': 1}
for word, freq in counter.items():
    if freq >= MIN_FREQ:
        vocab[word] = len(vocab)
vocab_size = len(vocab)

# 5. Padding: panjang maksimum = persentil ke-95 panjang tweet (padding di kiri
#    supaya state terakhir LSTM berasal dari token asli, bukan <PAD>)
max_len = int(np.percentile(train_df['tokens'].str.len(), 95))

def encode_pad(tokens, max_len):
    ids = [vocab.get(t, vocab['<UNK>']) for t in tokens][:max_len]
    return [vocab['<PAD>']] * (max_len - len(ids)) + ids

# 6. Konversi ke PyTorch Tensor
X_train = torch.tensor([encode_pad(t, max_len) for t in train_df['tokens']], dtype=torch.long)
y_train = torch.tensor(train_df['label'].values, dtype=torch.long)
X_test  = torch.tensor([encode_pad(t, max_len) for t in test_df['tokens']], dtype=torch.long)
y_test  = torch.tensor(test_df['label'].values, dtype=torch.long)

BATCH_SIZE = 32
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(TensorDataset(X_test, y_test), batch_size=BATCH_SIZE)

print(f'Ukuran vocabulary : {vocab_size}')
print(f'Panjang maksimum  : {max_len} token')
print(f'X_train: {tuple(X_train.shape)} | X_test: {tuple(X_test.shape)}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN LSTM</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432037_NiNyomanSutriani
class SequenceModel(nn.Module):
    def __init__(self, vocab_size, embed_dim=100, hidden_size=64, num_classes=3, dropout=0.5):
        super(SequenceModel, self).__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.lstm = nn.LSTM(embed_dim, hidden_size, batch_first=True)
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        emb = self.dropout(self.embedding(x))      # (batch, seq_len, embed_dim)
        _, (h_n, _) = self.lstm(emb)               # h_n: (1, batch, hidden_size)
        out = self.dropout(h_n[-1])                # hidden state terakhir
        return self.fc(out)                        # logits (batch, 3)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432037_NiNyomanSutriani
# 1. Inisiasi hyperparameter
epochs = 20
learning_rate = 0.001
embed_dim = 100
hidden_size = 64

# 2. Inisiasi model, loss (dengan class weight karena kelas tidak seimbang), dan optimizer
model = SequenceModel(vocab_size, embed_dim, hidden_size).to(device)

class_counts = np.bincount(y_train.numpy(), minlength=3)
class_weights = torch.tensor(len(y_train) / (3 * class_counts), dtype=torch.float).to(device)
criterion = nn.CrossEntropyLoss(weight=class_weights)
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Riwayat loss & akurasi
train_losses, test_losses, train_accs, test_accs = [], [], [], []

def evaluate(loader):
    model.eval()
    total_loss, correct, n = 0.0, 0, 0
    with torch.no_grad():
        for xb, yb in loader:
            xb, yb = xb.to(device), yb.to(device)
            out = model(xb)
            total_loss += criterion(out, yb).item() * len(yb)
            correct += (out.argmax(1) == yb).sum().item()
            n += len(yb)
    return total_loss / n, correct / n

# 4. Training loop
for epoch in range(1, epochs + 1):
    model.train()
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        loss = criterion(model(xb), yb)
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()

    tr_loss, tr_acc = evaluate(train_loader)
    te_loss, te_acc = evaluate(test_loader)
    train_losses.append(tr_loss); test_losses.append(te_loss)
    train_accs.append(tr_acc);    test_accs.append(te_acc)
    print(f'Epoch {epoch:02d}/{epochs} | Train Loss: {tr_loss:.4f} Acc: {tr_acc:.4f} | Test Loss: {te_loss:.4f} Acc: {te_acc:.4f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432037_NiNyomanSutriani
# 1. Plot Grafik Penurunan Loss Training terhadap Epoch
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.plot(test_losses, color='#1FA4B1', linewidth=2, linestyle='--', label='Test Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

# 2. Evaluasi klasifikasi: Prediksi vs Aktual pada data uji (confusion matrix)
model.eval()
all_preds = []
with torch.no_grad():
    for xb, _ in test_loader:
        all_preds.extend(model(xb.to(device)).argmax(1).cpu().tolist())
y_true = y_test.numpy()
labels = ['Negatif', 'Netral', 'Positif']

cm = confusion_matrix(y_true, all_preds)
plt.figure(figsize=(6, 5))
plt.imshow(cm, cmap='GnBu')
plt.colorbar()
plt.xticks(range(3), labels)
plt.yticks(range(3), labels)
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.title('Confusion Matrix (Data Uji)')
for i in range(3):
    for j in range(3):
        plt.text(j, i, cm[i, j], ha='center', va='center',
                 color='white' if cm[i, j] > cm.max() / 2 else '#114F6A', fontsize=13)
plt.show()

print(classification_report(y_true, all_preds, target_names=labels, digits=4))
```
